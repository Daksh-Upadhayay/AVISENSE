```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# Load data
column_names = ['engine_id', 'cycle', 'setting1', 'setting2', 'setting3', 'sensor1', 'sensor2', 'sensor3', 'sensor4', 'sensor5', 'sensor6', 'sensor7', 'sensor8', 'sensor9', 'sensor10', 'sensor11', 'sensor12', 'sensor13', 'sensor14', 'sensor15', 'sensor16', 'sensor17', 'sensor18', 'sensor19', 'sensor20', 'sensor21', 'sensor22', 'sensor23', 'sensor24', 'sensor25', 'sensor26']
train_df = pd.read_csv('/content/train_FD001.txt', sep='\s+', header=None, names=column_names)
test_df = pd.read_csv('/content/test_FD001.txt', sep='\s+', header=None, names=column_names)

# Drop columns with missing values
columns_to_drop = ['sensor22', 'sensor23', 'sensor24', 'sensor25', 'sensor26']
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

# Impute missing values in numerical features before scaling
numerical_features = [col for col in train_df.columns if col not in ['engine_id', 'cycle']]
imputer = SimpleImputer(strategy='mean')
train_df[numerical_features] = imputer.fit_transform(train_df[numerical_features])
test_df[numerical_features] = imputer.transform(test_df[numerical_features])

# Scale numerical features
scaler = StandardScaler()
train_df[numerical_features] = scaler.fit_transform(train_df[numerical_features])
test_df[numerical_features] = scaler.transform(test_df[numerical_features])

# Add rolling mean features
for sensor in ['sensor2', 'sensor3', 'sensor4', 'sensor7', 'sensor8', 'sensor9', 'sensor11', 'sensor12', 'sensor13', 'sensor14', 'sensor15', 'sensor17', 'sensor20', 'sensor21']:
    train_df[f'{sensor}_rolling_mean'] = train_df.groupby('engine_id')[sensor].transform(lambda x: x.rolling(window=10, min_periods=1).mean())
    test_df[f'{sensor}_rolling_mean'] = test_df.groupby('engine_id')[sensor].transform(lambda x: x.rolling(window=10, min_periods=1).mean())

# Impute missing values in rolling mean features using forward fill then backward fill
rolling_mean_features = [col for col in train_df.columns if '_rolling_mean' in col]
train_df[rolling_mean_features] = train_df[rolling_mean_features].fillna(method='ffill').fillna(method='bfill')
test_df[rolling_mean_features] = test_df[rolling_mean_features].fillna(method='ffill').fillna(method='bfill')


# Calculate RUL for training data
train_df['max_cycle'] = train_df.groupby('engine_id')['cycle'].transform('max')
train_df['RUL'] = train_df['max_cycle'] - train_df['cycle']
train_df = train_df.drop(columns=['max_cycle'])

# Calculate RUL for test data and merge with RUL_FD001.txt
rul_df = pd.read_csv('/content/RUL_FD001.txt', sep='\s+', header=None, names=['RUL'])
test_max_cycle = test_df.groupby('engine_id')['cycle'].max().reset_index()
test_max_cycle.columns = ['engine_id', 'max_cycle_test']
rul_df['engine_id'] = rul_df.index + 1
test_max_cycle = pd.merge(test_max_cycle, rul_df, on='engine_id')
test_df = pd.merge(test_df, test_max_cycle, on='engine_id', how='left')
test_df['RUL_test'] = test_df['RUL'] + (test_df['max_cycle_test'] - test_df['cycle'])

# Define failure threshold and create failure status
failure_threshold = 10
train_df['failure_status'] = (train_df['RUL'] <= failure_threshold).astype(int)
test_df['failure_status'] = (test_df['RUL_test'] <= failure_threshold).astype(int)


# Define features and target, excluding 'setting3' and 'sensor1'
features = [col for col in train_df.columns if col not in ['engine_id', 'cycle', 'RUL', 'failure_status', 'RUL_test', 'max_cycle_test', 'setting3', 'sensor1']]
X_train = train_df[features]
y_train = train_df['failure_status']
X_test = test_df[features]
y_test = test_df['failure_status']

# Train models with regularization parameters
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=10, min_samples_leaf=5) # Added regularization
gbm_model = GradientBoostingClassifier(n_estimators=100, learning_rate=0.05, max_depth=3, random_state=42) # Adjusted learning rate

rf_model.fit(X_train, y_train)
gbm_model.fit(X_train, y_train)

# Make predictions
rf_train_predictions = rf_model.predict(X_train)
gbm_train_predictions = gbm_model.predict(X_train)

rf_predictions = rf_model.predict(X_test)
gbm_predictions = gbm_model.predict(X_test)

# Evaluate models on training data
rf_train_accuracy = accuracy_score(y_train, rf_train_predictions)
rf_train_precision = precision_score(y_train, rf_train_predictions)
rf_train_recall = recall_score(y_train, rf_train_predictions)
rf_train_f1 = f1_score(y_train, rf_train_predictions)

gbm_train_accuracy = accuracy_score(y_train, gbm_train_predictions)
gbm_train_precision = precision_score(y_train, gbm_train_predictions)
gbm_train_recall = recall_score(y_train, gbm_train_predictions)
gbm_train_f1 = f1_score(y_train, gbm_train_predictions)


# Evaluate models on test data
rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_precision = precision_score(y_test, rf_predictions)
rf_recall = recall_score(y_test, rf_predictions)
rf_f1 = f1_score(y_test, rf_predictions)

gbm_accuracy = accuracy_score(y_test, gbm_predictions)
gbm_precision = precision_score(y_test, gbm_predictions)
gbm_recall = recall_score(y_test, gbm_predictions)
gbm_f1 = f1_score(y_test, gbm_predictions)

print("\nRandom Forest Model Evaluation (Training Data):")
print(f"Accuracy: {rf_train_accuracy:.4f}")
print(f"Precision: {rf_train_precision:.4f}")
print(f"Recall: {rf_train_recall:.4f}")
print(f"F1-score: {rf_train_f1:.4f}")

print("\nRandom Forest Model Evaluation (Test Data):")
print(f"Accuracy: {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall: {rf_recall:.4f}")
print(f"F1-score: {rf_f1:.4f}")


print("\nGradient Boosting Model Evaluation (Training Data):")
print(f"Accuracy: {gbm_train_accuracy:.4f}")
print(f"Precision: {gbm_train_precision:.4f}")
print(f"Recall: {gbm_train_recall:.4f}")
print(f"F1-score: {gbm_train_f1:.4f}")


print("\nGradient Boosting Model Evaluation (Test Data):")
print(f"Accuracy: {gbm_accuracy:.4f}")
print(f"Precision: {gbm_precision:.4f}")
print(f"Recall: {gbm_recall:.4f}")
print(f"F1-score: {gbm_f1:.4f}")

# Generate and display the confusion matrix for the gbm_model
conf_matrix_gbm = confusion_matrix(y_test, gbm_predictions)

plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix_gbm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Predicted 0', 'Predicted 1'], yticklabels=['Actual 0', 'Actual 1'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Gradient Boosting Model')
plt.show()
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Assuming new_data_df is loaded and preprocessed before this cell

# Load and preprocess new data
# Assuming new_data_df is loaded from a file named 'engine_failure_dataset.csv'
new_data_df = pd.read_csv('/content/engine_failure_dataset.csv')

new_data_df['Time_Stamp'] = pd.to_datetime(new_data_df['Time_Stamp'])
start_time = new_data_df['Time_Stamp'].min()
new_data_df['time_in_seconds'] = (new_data_df['Time_Stamp'] - start_time).dt.total_seconds()
new_data_df['failure_status_new'] = (new_data_df['Fault_Condition'] != 0).astype(int)
numerical_features_new = ['Temperature (°C)', 'RPM', 'Fuel_Efficiency', 'Vibration_X', 'Vibration_Y', 'Vibration_Z', 'Torque', 'Power_Output (kW)']
scaler_new = StandardScaler()
new_data_df[numerical_features_new] = scaler_new.fit_transform(new_data_df[numerical_features_new])
for feature in numerical_features_new:
    new_data_df[f'{feature}_rolling_mean'] = new_data_df[feature].transform(lambda x: x.rolling(window=10, min_periods=1).mean())


# Define features and target for the new data
features_new = [col for col in new_data_df.columns if col not in ['Time_Stamp', 'Fault_Condition', 'failure_status_new', 'Operational_Mode']]
X_new = new_data_df[features_new]
y_new = new_data_df['failure_status_new']

# Train the new Gradient Boosting Model
gbm_new = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
gbm_new.fit(X_new, y_new)

# Make predictions
new_data_df['predicted_failure_status'] = gbm_new.predict(X_new)
display(new_data_df.head())

# Generate and display the confusion matrix for the gbm_new model
conf_matrix_gbm_new = confusion_matrix(new_data_df['failure_status_new'], new_data_df['predicted_failure_status'])

plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix_gbm_new, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Predicted 0', 'Predicted 1'], yticklabels=['Actual 0', 'Actual 1'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for New Gradient Boosting Model (gbm_new)')
plt.show()
```
